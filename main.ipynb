```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Numerical Variables")
```

```python
sns.scatterplot(data=df, x="PM2.5", y="AQI")
plt.title("PM2.5 vs AQI")
```
