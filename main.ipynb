```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
missing_count = df.isnull().sum()
missing_count
```

```python
missing_count_percentage = (df.isnull().sum() / len(df)) * 100

missing_count_percentage.sort_values(ascending=False)
```

```python
missing_count_percentage = missing_count_percentage[missing_count_percentage > 0]
plt.figure(figsize=(12,8))

missing_count_percentage.sort_values(ascending=False).plot(kind="bar")

plt.xlabel("Feature")
plt.ylabel("Missing_values")
plt.title("Percentage of Missing avlues")
plt.xticks(rotation=45)
plt.show()
```

```python
dupplicated_count  = df.duplicated().sum()
dupplicated_count
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

negative_column = df[df[numeric_columns] < 0 ].sum()

negative_column
```

```python
zero_count = (df[numeric_columns] == 0).sum()
zero_count
```

```python
plt.figure(figsize=(14, 8))

sns.boxplot(
    data=df[numeric_columns],
    orient="h"
)

plt.title("Distribution and Potential Outliers of Numerical Variables")
plt.xlabel("Value")

plt.show()
```

```python
pollutant_columns = [
    "PM2.5",
    "PM10",
    "NO",
    "NO2",
    "NOx",
    "NH3",
    "CO",
    "SO2",
    "O3",
    "Benzene",
    "Toluene",
    "Xylene"
]
```

```python
for column in pollutant_columns:
    if column in df.columns:
        plt.figure(figsize=(8, 4))
        sns.boxplot(x=df[column])
        plt.title(f"Boxplot of {column}")
        plt.xlabel(column)
        plt.show()
```
